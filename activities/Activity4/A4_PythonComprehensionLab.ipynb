**The problem:** given a list of device battery levels (0–100), decide what action each device needs.

## Section 1 — Imperative Style

The imperative style solves the problem step by step, using a loop and explicit `if`/`elif`/`else` branches that build up a result as the program runs.

```python
battery_levels = [5, 45, 82, 19, 100, 0, 63]

actions_imperative = []
for level in battery_levels:
    if level < 20:
        action = "Critical - recharge now"
    elif level < 50:
        action = "Low - recharge soon"
    else:
        action = "OK"
    actions_imperative.append(action)

for level, action in zip(battery_levels, actions_imperative):
    print(f"Battery {level}%: {action}")
```

## Section 2 — Functional Style

The functional style avoids step-by-step state changes. Instead, it defines a function and applies it across the whole list using `map`, and uses `filter` to select items matching a condition.

```python
def classify_battery(level):
    if level < 20:
        return "Critical - recharge now"
    elif level < 50:
        return "Low - recharge soon"
    else:
        return "OK"

actions_functional = list(map(classify_battery, battery_levels))
print(actions_functional)

critical_devices = list(filter(lambda level: level < 20, battery_levels))
print("Devices needing urgent attention:", critical_devices)
```

## Section 3 — Logic / Rule-Based Style

The logic/rule-based style — the flavor Session 6 connects to expert systems — expresses the solution as a table of **rules**: a list of `(condition, result)` pairs, checked in order until one matches. This looks more like declarative "knowledge" than a sequence of instructions.

```python
battery_rules = [
    (lambda level: level < 20, "Critical - recharge now"),
    (lambda level: level < 50, "Low - recharge soon"),
    (lambda level: True,       "OK"),
]

def apply_rules(level, rules):
    for condition, result in rules:
        if condition(level):
            return result
    return "No rule matched"

actions_logic = [apply_rules(level, battery_rules) for level in battery_levels]
print(actions_logic)
```
